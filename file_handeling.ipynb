```python
f1 = open("data.txt","w")
#write- adds a string
 #writelines-to write multiple lines
content="in a new file"
f1.write(content)
print ("file created")
f1.close()
```

```python
f1=open("data.txt","a")
f1.writelines(["\n3rd line","\n4th line"])
f1.close()
```

```python
with open("data.txt","r") as f1:
  content=f1.read(10)
  print(content)
  print(f1.tell())
  f1.seek(2)
  print(f1.read())
```

```python
with open("data.txt","r") as f1:
  content=f1.readlines()
  print(content)
```
