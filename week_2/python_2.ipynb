```python
#greatest among three numbers
a=int(input("enter a value:"))
b=int(input("enter a value:"))
c=int(input("enter a value:"))
if(a>b and a>c):
  print(f"{a}is greater")
elif(b>c and b>a):
    print("f{b}is greater")
else:
    print(f"{c} is greater")

```

```python
#print FIZZ if a number divisible by 3 , print BUZZ if divisible by 5,print FIZZBUZZ if divisible by both 3 and 5
a=int (input("enter a value"))
if(a%3==0):
   print("FIZZ",end="")
if(a%5==0):
   print("BUZZ")
```

```python
#print FIZZ if a number divisible by only 3 , print BUZZ if divisible by 5 only,print FIZZBUZZ if divisible by both 3 and 5
a=int (input("enter a value"))
if(a%3==0 and a%5!=0):
   print("FIZZ")
elif(a%5==0 and a%3!=0):
   print("BUZZ")
elif(a%3==0 and a%5==0):
   print("FIZZBUZZ")
else:
  print ("invalid")
```

```python
#program on match
num=int(input("enter a number:"))
match num:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
  case _:
    print("invalid")
```
