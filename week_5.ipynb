```python
#program that count the no of vowels and consonents in a string
n="hello world"
count1=0
count2=0
for j in n:
  if j=="a" or j=="e" or j=="i" or j=="o" or j=="u" :
    count1=count1+1
  else:
    count2=count2+1
print(count1)
print(count2)
```

```python
a=str(input())
b=a.lower()
vowels="aeiou"
count_v=0
count_c=0
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      count_v+=1
    else:
      count_c+=1
print(count_v)
print(count_c)
```
